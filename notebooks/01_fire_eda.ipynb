```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
project = Path.home() / "Documents" / "Wildfire Emergency Response Analytics"
csv_files = list(project.rglob("*.csv"))

print("CSV files found:", len(csv_files))

for file in csv_files:
    print(file)
csv_file = csv_files[0]
df = pd.read_csv(csv_file)

print("\nLoaded:", csv_file.name)
print("Rows:", len(df))
print("Columns:", len(df.columns))
df_clean = df.copy()
df_clean = df_clean.drop_duplicates()
df_clean["acq_date"] = pd.to_datetime(
    df_clean["acq_date"],
    errors="coerce"
)
df_clean["year"] = df_clean["acq_date"].dt.year
df_clean["month"] = df_clean["acq_date"].dt.month
df_clean["month_name"] = df_clean["acq_date"].dt.month_name()

print("\nClean dataset ready!")
print("Rows after cleaning:", len(df_clean))

df_clean.head()
```

```python
print(df_clean.columns.tolist())
```

```python
plt.figure(figsize=(10, 5))

plt.hist(
    df_clean["frp"].dropna(),
    bins=30
)

plt.title("Fire Radiative Power (FRP) Distribution")
plt.xlabel("FRP")
plt.ylabel("Number of Fire Detections")

plt.tight_layout()
plt.show()
```

```python
top_frp = (
    df_clean[
        ["acq_date", "latitude", "longitude", "brightness", "frp"]
    ]
    .sort_values("frp", ascending=False)
    .head(10)
    .reset_index(drop=True)
)

top_frp.index = top_frp.index + 1

top_frp
```

```python
df_clean["fire_intensity"] = pd.cut(
    df_clean["frp"],
    bins=[0, 50, 100, 300, float("inf")],
    labels=["Low", "Medium", "High", "Extreme"],
    include_lowest=True
)

intensity_counts = (
    df_clean["fire_intensity"]
    .value_counts()
    .reindex(
        ["Low", "Medium", "High", "Extreme"],
        fill_value=0
    )
)

intensity_counts
```

```python
plt.figure(figsize=(8, 5))

intensity_counts.plot(kind="bar")

plt.title("Wildfire Detection by Fire Intensity")
plt.xlabel("Fire Intensity")
plt.ylabel("Number of Detections")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
monthly_intensity = (
    df_clean.groupby(
        ["month", "month_name", "fire_intensity"],
        observed=True
    )
    .size()
    .reset_index(name="fire_count")
)

monthly_intensity.head(20)
```

```python
high_extreme = (
    df_clean[
        df_clean["fire_intensity"].isin(
            ["High", "Extreme"]
        )
    ]
    .groupby(["month", "month_name"])
    .size()
    .reset_index(name="high_intensity_fires")
    .sort_values("month")
)

high_extreme
```

```python
plt.figure(figsize=(10, 5))

plt.bar(
    high_extreme["month_name"],
    high_extreme["high_intensity_fires"]
)

plt.title("High & Extreme Wildfire Detections by Month")
plt.xlabel("Month")
plt.ylabel("High / Extreme Fire Detections")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
peak_intensity = high_extreme.loc[
    high_extreme["high_intensity_fires"].idxmax()
]

print(
    "Month with highest high/extreme fires:",
    peak_intensity["month_name"]
)

print(
    "High/Extreme fire detections:",
    peak_intensity["high_intensity_fires"]
)
```

```python

df_clean["lat_grid"] = df_clean["latitude"].round(1)
df_clean["lon_grid"] = df_clean["longitude"].round(1)

hotspots = (
    df_clean.groupby(
        ["lat_grid", "lon_grid"]
    )
    .size()
    .reset_index(name="fire_count")
    .sort_values(
        "fire_count",
        ascending=False
    )
)

hotspots.head(10)
```

```python
top_hotspots = hotspots.head(10).reset_index(drop=True)

top_hotspots.index = top_hotspots.index + 1

top_hotspots
```

```python
plt.figure(figsize=(12, 6))

plt.scatter(
    hotspots["lon_grid"],
    hotspots["lat_grid"],
    s=hotspots["fire_count"] * 2,
    alpha=0.5
)

plt.title("Global Wildfire Detection Hotspots - 2025")
plt.xlabel("Longitude")
plt.ylabel("Latitude")

plt.grid(True, alpha=0.2)

plt.tight_layout()
plt.show()
```

```python
top_hotspot = hotspots.iloc[0]

print("🔥 Highest Fire Hotspot")
print("----------------------")
print("Latitude :", top_hotspot["lat_grid"])
print("Longitude:", top_hotspot["lon_grid"])
print("Fire detections:", top_hotspot["fire_count"])
```

```python
from pathlib import Path

output_folder = (
    project / "data" / "processed"
)

output_folder.mkdir(
    parents=True,
    exist_ok=True
)

output_file = (
    output_folder / "wildfire_analytics_2025.csv"
)

df_clean.to_csv(
    output_file,
    index=False
)

print("✅ Processed dataset saved!")
print(output_file)
```

```python
total_detections = len(df_clean)

high_fires = (
    df_clean["fire_intensity"].isin(["High", "Extreme"])
).sum()

extreme_fires = (
    df_clean["fire_intensity"] == "Extreme"
).sum()

avg_frp = df_clean["frp"].mean()
max_frp = df_clean["frp"].max()

print("========== WILDFIRE RESPONSE KPIs ==========")
print(f"Total Fire Detections : {total_detections:,}")
print(f"High/Extreme Fires    : {high_fires:,}")
print(f"Extreme Fires         : {extreme_fires:,}")
print(f"Average FRP           : {avg_frp:.2f}")
print(f"Maximum FRP           : {max_frp:.2f}")
```

```python
severe_percentage = (
    high_fires / total_detections * 100
)

extreme_percentage = (
    extreme_fires / total_detections * 100
)

print(
    f"High/Extreme Fire Rate: {severe_percentage:.2f}%"
)

print(
    f"Extreme Fire Rate: {extreme_percentage:.2f}%"
)
```

```python
daily_fires = (
    df_clean.groupby("acq_date")
    .size()
    .reset_index(name="fire_count")
)

daily_fires.head()
```

```python
peak_day = daily_fires.loc[
    daily_fires["fire_count"].idxmax()
]

print("🔥 Peak Fire Detection Day")
print("--------------------------")
print("Date:", peak_day["acq_date"].date())
print("Fire detections:", peak_day["fire_count"])
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    daily_fires["acq_date"],
    daily_fires["fire_count"]
)

plt.title("Daily Wildfire Detection Trend - 2025")
plt.xlabel("Date")
plt.ylabel("Fire Detections")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
kpi_summary = pd.DataFrame({
    "Metric": [
        "Total Fire Detections",
        "High/Extreme Fires",
        "Extreme Fires",
        "Average FRP",
        "Maximum FRP",
        "High/Extreme Fire Rate (%)",
        "Extreme Fire Rate (%)"
    ],
    "Value": [
        total_detections,
        high_fires,
        extreme_fires,
        round(avg_frp, 2),
        round(max_frp, 2),
        round(severe_percentage, 2),
        round(extreme_percentage, 2)
    ]
})

kpi_summary
```

```python
kpi_summary.to_csv(
    output_folder / "wildfire_kpi_summary.csv",
    index=False
)

print("✅ KPI summary saved!")
```
